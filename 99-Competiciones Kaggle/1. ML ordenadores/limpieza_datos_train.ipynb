```python
# Touchscreen: 1 si tiene, 0 si no (que la pantalla sea táctil)
df['Touchscreen'] = df['ScreenResolution'].str.contains('Touchscreen').astype(int)

# IPS Panel: 1 si tiene, 0 si no (IPS más caro, temas de color y ángulos de la pantalla)
df['IPS'] = df['ScreenResolution'].str.contains('IPS').astype(int)

# Resolución: extraer solo los píxeles
df['Resolution'] = df['ScreenResolution'].str.extract(r'(\d+x\d+)')
```

```python
# Velocidad en GHz
df['CPU_GHz'] = df['Cpu'].str.extract(r'(\d+\.?\d*)GHz').astype(float)

# Marca: Intel (1) o AMD (0) 
df['CPU_brand'] = df['Cpu'].apply(lambda x: x.split()[0]) # coge si es AMD o Intel (primera palabra)
df['CPU_brand'] = le.fit_transform(df['CPU_brand']) # aquí aplica 0 a AMD y 1 a Intel
```

```python
# diccionario de número ascendente por categoría de procesador
cpu_order = {
    # Intel gama baja
    'Atom': 0, 'Celeron': 1, 'Pentium': 2,
    # AMD gama baja
    'E-Series': 3, 'A4-Series': 4, 'A6-Series': 5, 'A8-Series': 6,
    'A9-Series': 7, 'A10-Series': 8, 'A12-Series': 9,
    # Intel gama media-alta
    'Core M': 10, 'Core i3': 11, 'Core i5': 12, 'Core i7': 13,
    # Gama profesional/alta
    'Xeon': 14, 'Ryzen': 15
    }

# establecer los números a cada categoría

def extract_cpu_model(cpu):
    parts = cpu.split()
    # Intel Core i5/i7/i3/M necesita 2 palabras
    if parts[1] == 'Core':
        return parts[1] + ' ' + parts[2]
    # resto con 1 palabra
    return parts[1]

df['CPU_model'] = df['Cpu'].apply(extract_cpu_model)
df['CPU_model'] = df['CPU_model'].map(cpu_order)
```

```python
# separar tres columnas: SSD, HDD, Hybrid

def extract_storage(memory):
    ssd = 0
    hdd = 0
    hybrid = 0
    parts = memory.split('+')  # divide por '+' por si hay dos almacenamientos: ['256GB SSD', '1TB HDD']
    for part in parts:
        part = part.strip()  # quita espacios en blanco sobrantes
        
        # extrae el número en GB
        if 'TB' in part:
            size = int(float(part.split('TB')[0].strip()) * 1024)  # convierte TB a GB: 1TB → 1024GB
        else:
            size = int(float(part.split('GB')[0].strip()))  # ya está en GB: 256GB → 256
        
        # clasifica en SSD, HDD o Hybrid según las palabras clave
        if any(x in part for x in ['SSD', 'Flash']):
            ssd += size      # += acumula por si hay varios SSD: 512GB SSD + 256GB SSD → 768GB
        elif 'HDD' in part:
            hdd += size      # += acumula por si hay varios HDD: 1TB HDD + 1TB HDD → 2048GB
        elif 'Hybrid' in part:
            hybrid += size   # += acumula por si hay varios Hybrid

    return pd.Series([ssd, hdd, hybrid])  # devuelve los 3 valores a la vez

# asigna los 3 valores directamente a 3 columnas nuevas
df[['SSD_GB', 'HDD_GB', 'Hybrid_GB']] = df['Memory'].apply(extract_storage)
df.drop('Memory', axis=1, inplace=True)  # elimina la columna original ya que no la necesitamos
```

```python
import re

def extract_gpu_numbers(gpu):
    intel = 0
    amd = 0
    nvidia = 0  # iniciamos las 3 columnas a 0, si no tiene esa marca se queda en 0
    
    # extraer el primer número que aparece en el string
    number = re.search(r'(\d+)', gpu)  # busca el primer número en el string: 'GTX 1080' → 1080
    if number:
        number = int(number.group(1))  # lo convierte a int: '1080' → 1080
    else:
        number = 0  # si no hay número (ej: 'Intel HD Graphics') → 0
    
    # asigna el número a la columna de su marca, las otras se quedan en 0
    if 'Intel' in gpu:
        intel = number
    elif 'AMD' in gpu:
        amd = number
    elif 'Nvidia' in gpu:
        nvidia = number
    
    return pd.Series([intel, amd, nvidia])  # devuelve los 3 valores a la vez

# asigna los 3 valores directamente a 3 columnas nuevas
df[['GPU_Intel', 'GPU_AMD', 'GPU_Nvidia']] = df['Gpu'].apply(extract_gpu_numbers)
df.drop('Gpu', axis=1, inplace=True)  # elimina la columna original ya que no la necesitamos
```
