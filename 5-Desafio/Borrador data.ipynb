# MUSEOS Y PATRIMONIO ENRIQUECIDO

```python
df_def = pd.read_csv("museos_patrimonio_enriquecido.csv")
df_def
```

```python
df_def.drop(df_def[df_def['Nombre'].str.contains('ayuntamiento', case=False, na=False)].index, inplace=True)
df_def.drop(df_def[df_def['Nombre'].str.contains('casco', case=False, na=False)].index, inplace=True)
df_def.drop(df_def[df_def['Nombre'].str.contains('ciudad', case=False, na=False)].index, inplace=True)
df_def.drop(df_def[df_def['Nombre'].str.contains('parte', case=False, na=False)].index, inplace=True)
df_def.drop(df_def[df_def['Nombre'].str.contains('iñigo', case=False, na=False)].index, inplace=True)
df_def.drop(df_def[df_def["Tipo de lugar"] == "Artesanos y asociaciones"].index, inplace=True)
df_def.drop(df_def[df_def["valoracion"] < 4].index, inplace=True)
df_def.drop(columns=['Marcas'], inplace=True)
df_def.drop(columns=['Coordenadas GPS'], inplace=True)
```

```python
df_def["Tipo de lugar"].unique()
```

```python
df_def['Tipo de lugar'] = df_def['Tipo de lugar'].replace({
    'Auditorios; Museos; Galerías de arte y salas de exposiciones': 'Museo',
    'Museos' : 'Museo',
    'Teatros y Cines': 'Patrimonio cultural',
    'Puertos pesqueros': 'Patrimonio cultural'
})
```

```python
df_def = df_def.rename(columns={
    'LATWGS84' : 'lat',
    'LONWGS84' : 'lon'
})
```

```python
df_def['Tipo de Cultura'] = df_def['Tipo de Cultura'].fillna('Otros')
```

# POST LIMPIEZA EN EXCEL

```python
df_def = pd.read_excel("Libro1.xlsx")
df_def
```

```python
df_def['valoracion'] = df_def['valoracion'] / 10
df_def
```

```python
# LAT: 2 números antes del punto
df_def['lat'] = df_def['lat'].astype(str).str.replace('.', '', regex=False)
df_def['lat'] = df_def['lat'].str[:2] + '.' + df_def['lat'].str[2:]

# LON: 1 número antes del punto (manteniendo el signo -)
df_def['lon'] = df_def['lon'].astype(str).str.replace('.', '', regex=False)

df_def['lon'] = df_def['lon'].apply(
    lambda x: '-' + x[1] + '.' + x[2:] if x.startswith('-')
    else x[0] + '.' + x[1:]
)

# convertir otra vez a float
df_def['lat'] = df_def['lat'].astype(float)
df_def['lon'] = df_def['lon'].astype(float)
```
